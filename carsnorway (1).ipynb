```python
#year comparison between sales of new cars to used cars in Norway
car_sales_trends = combined_df.groupby('Year').agg(
    total_new_cars=('Quantity', 'sum'),
    total_used_cars=('Used', 'sum')
).reset_index()
car_sales_trends
```

```python
#the ratio between the used and new car. 
car_sales_trends['used_to_new_ratio'] = car_sales_trends['total_used_cars'] / car_sales_trends['total_new_cars']
print(car_sales_trends[['Year', 'total_new_cars', 'total_used_cars', 'used_to_new_ratio']])
```

```python
plt.title('Used-to-New Car Ratio by Year')
plt.xlabel('Year')
plt.ylabel('used_to_new_ratio')
plt.grid(True)  # Add grid for better readability
plt.xticks(car_sales_trends['Year'])  # Ensuring all years are shown on the x-axis
plt.tight_layout()  # Adjust layout for better appearance

# Show the chart
plt.show()
```

```python
# Group by Year and Month to calculate total sales and mean YoY changes
sales_trends = combined_df.groupby(['Year', 'Month']).agg(
    total_sales=('Quantity', 'sum'),
    avg_quantity_yoy=('Quantity_YoY', 'mean'),
    avg_import_yoy=('Import_YoY', 'mean')
)
top_5_sales = sales_trends.sort_values(by='total_sales', ascending=False).head(5)
top_5_sales
```

```python
# Aggregate sales trends by Year
yearly_trends = sales_trends.groupby('Year').agg(
    total_sales=('total_sales', 'sum'),
    avg_quantity_yoy=('avg_quantity_yoy', 'mean'),
    avg_import_yoy=('avg_import_yoy', 'mean')
).reset_index()

# Plotting
plt.figure(figsize=(6, 4))
plt.plot(yearly_trends['Year'], yearly_trends['total_sales'], label='Total Sales Quantity', color='blue', marker='o')
plt.plot(yearly_trends['Year'], yearly_trends['avg_quantity_yoy'], label='Quantity YoY (%)', color='red', linestyle='--')
plt.plot(yearly_trends['Year'], yearly_trends['avg_import_yoy'], label='Import YoY (%)', color='green', linestyle='--')

# Adding labels, title, and grid
plt.xlabel('Year')
plt.ylabel('Quantity / YoY (%)')
plt.title('Trends in Car Sales Quantity and YoY Changes')
plt.legend()

# Show the plot
plt.tight_layout()
plt.show()
```

```python
# streamlit_dashboard.py

import streamlit as st
import matplotlib.pyplot as plt
import pandas as pd

# `combined_df`
st.title("Norway Car Sales Analysis")

# Top 20 Models Share Plot
st.subheader("Top 20 Models Performance in Total Car Sales")
fig1, ax1 = plt.subplots()
ax1.plot(combined_df['Year'], combined_df['sales percentage'], marker='o', color='b', label='Top 20 Models Share (%)')
ax1.set_xlabel('Year')
ax1.set_ylabel('Percentage of Total Sales')
ax1.legend()
st.pyplot(fig1)

# Fuel Type Distribution
st.subheader("Distribution of Different Fuel Types in Car Sales")
fuel_cars_total = {
    'Electric': combined_df['Quantity_Electric'].sum(),
    'Hybrid': combined_df['Quantity_Hybrid'].sum(),
    'Diesel': combined_df['Quantity_Diesel'].sum()
}
fig2, ax2 = plt.subplots()
ax2.bar(fuel_cars_total.keys(), fuel_cars_total.values(), color=['blue', 'red', 'green'])
ax2.set_xlabel('Fuel Type')
ax2.set_ylabel('Total Quantity')
st.pyplot(fig2)

# Top 5 Car Makes by Total Sales
st.subheader("Top 5 Car Makes by Total Sales")
top_cars = combined_df.groupby('Make').agg({
    'Quantity_Electric': 'sum',
    'Quantity_Diesel': 'sum',
    'Quantity_Hybrid': 'sum',
})
top_cars['total_sales'] = top_cars.sum(axis=1)
top_5_cars = top_cars.sort_values(by='total_sales', ascending=False).head(5)
st.table(top_5_cars)
```
